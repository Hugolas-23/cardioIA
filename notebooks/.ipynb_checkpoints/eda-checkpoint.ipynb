```python
df_sem = df.copy()
```

```python
df_sem.drop(columns=['ca'], inplace=True)
```

```python
df.columns
```

```python
df_sem["thal"] = df_sem.groupby(["cp", "sex"])["thal"].transform(lambda x: x.fillna(x.mode()[0]))
#faz o imput nos valores faltantes de thal de acordo com o grupo cp e sex
```

```python
df_sem.isnull().sum()
```

```python
df_sem['slope'] = df_sem.groupby(["cp", "sex"])["slope"].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df_sem.isnull().sum()
```

```python
df_sem['trestbps'] = df_sem.groupby(["cp", "sex"])["trestbps"].transform(lambda x: x.fillna(x.median()))
df_sem.isnull().sum()
```

```python
df_sem['chol'] = df_sem.groupby(["cp", "sex"])["chol"].transform(lambda x: x.fillna(x.median()))
df_sem.isnull().sum()
```

```python
df_sem['fbs'] = df_sem.groupby(["cp", "sex"])["fbs"].transform(lambda x: x.fillna(x.mode()[0]))
df_sem.isnull().sum()
```

```python
df_sem['restecg'] = df_sem.groupby(["cp", "sex"])["restecg"].transform(lambda x: x.fillna(x.mode()[0]))
df_sem.isnull().sum()
```

```python
df_sem['thalch'] = df_sem.groupby(["cp", "sex"])["thalch"].transform(lambda x: x.fillna(x.median()))
df_sem.isnull().sum()
```

```python
df_sem['exang'] = df_sem.groupby(["cp", "sex"])["exang"].transform(lambda x: x.fillna(x.mode()[0]))
df_sem.isnull().sum()
```

```python
df_sem['oldpeak'] = df_sem.groupby(["cp", "sex"])["oldpeak"].transform(lambda x: x.fillna(x.median()))
df_sem.isnull().sum()
```

```python
#binarizando a coluna "num"
df_sem["target"] = (df_sem["num"] > 0).astype(int)
print(df_sem["target"].value_counts())
```

```python
df_sem = df_sem.drop(columns = ["num"])
df_sem
```
